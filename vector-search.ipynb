## Splitting the books using CharacterTextSplitter

```python
# Creating text document for CharacterTextSplitter

books['tagged_description'].to_csv('tagged_description.txt',
                                   index=False,
                                   header=False)
```

```python
raw_documents = TextLoader("tagged_description.txt").load()
text_splitter = CharacterTextSplitter(chunk_size=1, chunk_overlap=0, separator='\n')
documents = text_splitter.split_documents(raw_documents)
```

```python
documents[0]
```

## Building vector databases

Creating document embeddings and storing in vector database

```python
embedding_model = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
db_books = Chroma.from_documents(
    documents,
    embedding=embedding_model
)
```

## Getting book recommendations using vector search

```python
query = "a book about a young wizard who discovers his magical heritage and attends a school for wizards, where he makes friends and uncovers the truth about his past."
docs = db_books.similarity_search(query, k=10)
docs
```

```python
books[books["isbn13"]== int(docs[0].page_content.split()[0].strip('"'))]
```

```python
def retrieve_semantic_recommendaions(
        query:str,
        top_k:int=10
)-> pd.DataFrame:
    recs = db_books.similarity_search(query, k=50)
    book_list = []
    # basic loop
    # for i in range(0, len(recs)):
    #     book_list += [int(recs[i].page_content.strip('"').split()[0])]
    
    # List comprehension
    # book_list = [int(recs[i].page_content.strip('"').split()[0]) for i in range(0, len(recs))]
    
    # build in: map function
    book_list = list(map(lambda x: int(x.page_content.strip('"').split()[0]), recs))

    return books[books["isbn13"].isin(book_list)].head(top_k)
```

```python
retrieve_semantic_recommendaions("a book about a young wizard who discovers his magical heritage and attends a school for wizards, where he makes friends and uncovers the truth about his past.", top_k=10)
```

## Simplifying categories

```python
category_mapping = {
    'Fiction': 'Fiction',
    'Juvenile Fiction': "Children's Fiction",
    'Biography & Autobiography': 'Nonfiction',
    'History': 'Nonfiction',
    'Literary Criticism': 'Nonfiction',
    'Philosophy': 'Nonfiction',
    'Religion': 'Nonfiction',
    'Comics & Graphic Novels': 'Fiction',
    'Drama': 'Fiction',
    'Juvenile Nonfiction': "Children's Nonfiction",
    'Science': 'Nonfiction',
    'Poetry': 'Fiction',
}

books["simple_categories"] = books["categories"].map(category_mapping)
```

```python
books[~(books['simple_categories'].isna())]
```

## Zero-shot classification

```python
from transformers import pipeline

fiction_categories=['Fiction', 'Nonfiction']
pipe = pipeline("zero-shot-classification", model="facebook/bart-large-mnli", device=-1)
# result = classifier(some_description, candidate_labels=["fiction", "nonfiction"])
```

```python
sequence = books.loc[books['simple_categories']=="Fiction", 'description'].reset_index(drop=True)[0]
```

```python
pipe(sequence, fiction_categories)
```

```python
import numpy as np

max_index=np.argmax(pipe(sequence, fiction_categories)['scores'])
max_label=pipe(sequence, fiction_categories)['labels'][max_index]
max_label
```

```python
def generate_prediction(sequence:str, category)->str:
    predictions = pipe(sequence, category)
    max_index=np.argmax(predictions['scores'])
    max_label=predictions['labels'][max_index]
    return max_label
```

```python

actual_cats = []
predicted_cats =[]
```

```python
from tqdm import tqdm

for i in tqdm(range(0, 300)):
    sequence = books.loc[books['simple_categories']=="Fiction",'description'].reset_index(drop=True)[i]
    predicted_cats += [generate_prediction(sequence, fiction_categories)]
    actual_cats+=["Fiction"]
```

```python
for i in tqdm(range(0, 300)):
    sequence = books.loc[books['simple_categories']=='Nonfiction', 'description'].reset_index(drop=True)[i]
    predicted_cats += [generate_prediction(sequence, fiction_categories)]
    actual_cats+=["Nonfiction"]
```

```python
predictions_df = pd.DataFrame({'actual_categories': actual_cats, 'predicted_categories': predicted_cats})
```

```python
predictions_df['correct_prediction'] = (
    np.where(predictions_df['actual_categories'] == predictions_df['predicted_categories'], 1, 0)   
)
predictions_df['correct_prediction'].sum()/len(predictions_df)
```

```python
isbns = []
predicted_cats=[]

missing_cats = books.loc[books['simple_categories'].isna(), ['isbn13','description']].reset_index(drop=True)
```

```python
for i in tqdm(range(0, len(missing_cats))):
    sequence = missing_cats.loc[i, 'description']
    predicted_cats += [generate_prediction(sequence, fiction_categories)]
    isbns += [missing_cats.loc[i, 'isbn13']]
```

```python

missing_predicted_df= pd.DataFrame({'isbn13':isbns, 'predicted_cats':predicted_cats})
```

```python
len(isbns), len(predicted_cats), len(missing_cats)
```

```python
missing_predicted_df.to_csv('missing_predicted.csv', index=False)
```

```python
books =pd.merge(books, missing_predicted_df, on='isbn13', how='left')
prediction_cols = [
    col for col in books.columns
    if col == "predicted_cats" or col.startswith("predicted_cats_")
]
books = books.drop(columns=prediction_cols)

books = books.merge(missing_predicted_df, on="isbn13", how="left")
books["simple_categories"] = books["simple_categories"].fillna(
    books["predicted_cats"]
)
books = books.drop(columns="predicted_cats")
```
